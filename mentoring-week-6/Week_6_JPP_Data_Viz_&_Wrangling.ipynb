## Detail function

- The function only needs one input, `filenames` (`list`), a list of transactions data files.
- The `import_data` function will join every data listed on the filenames as a Pandas DataFrame.

## Answer
---


- Provide the code for solving the problem
- **Make sure your function follows the `Detail Function`**

`````python
def import_data(filenames):
    """
    Takes multiple files with .xlsx and .csv extensions and combined all of
    them into a single pandas.DataFrame.
    """
    from pathlib import Path
    dfs = []
    idl = []
    for name in filenames:
        # export the extension    
        ext = Path(name).suffix
        # store the read data frame inside a list
        if ext == ".xlsx":
            data = pd.read_excel(name)
        else:
            data = pd.read_csv(name, delimiter=';')

        dfs.append(data)
        # store index length of each data frame for later verification
        idl.append(len(data.index))
    
    # concat all the list available using pd.concat()
    main_data = pd.concat(dfs)

    # verify the data by totalling the index length
    if len(main_data.index) == sum(idl):
        return main_data
    else:
        raise ImportError('There is something wrong about the import process')

filenames = ['branch_A.xlsx', 'branch_B.csv', 'branch_C.csv']
data = import_data(filenames=filenames)
data.shape
`````

- Provide the code for solving the problem

`````python
CONFIG = {
    'path': {
        'user_data': 'ratings.csv',
        'metadata': 'movies.csv'
    }
}

def get_unwatched_movie(userId, config=CONFIG):
    # load data
    user_data = pd.read_csv(config["path"]["user_data"])
    metadata = pd.read_csv(config["path"]["metadata"])
    # select unique watched movieId
    watched_movieIds = user_data[user_data['userId'] == userId]['movieId'].unique()
    # separate unwatched movies from the watched ones
    unwatched_movies = metadata[~metadata['movieId'].isin(watched_movieIds)]
    unwatched_movies.set_index('movieId', inplace=True)
    return unwatched_movies    
`````

## Answer
---

- Provide the code for solving the problem
- **Make sure your function follows the `Detail Function`**

`````python
import math

# haversine function
def haversine(lat1, lon1, lat2, lon2):
    """
    https://www.geeksforgeeks.org/haversine-formula-to-find-distance-between-two-points-on-a-sphere/
    """
    dLat = (lat2 - lat1) * math.pi / 180.0
    dLon = (lon2 - lon1) * math.pi / 180.0
    lat1 = lat1 * math.pi / 180.0
    lat2 = lat2 * math.pi / 180.0
    a = (pow(math.sin(dLat/2), 2) + pow(math.sin(dLon/2), 2) * math.cos(lat1) * math.cos(lat2))
    rad = 6371 # earth radius in km
    c = 2 * math.asin(math.sqrt(a))
    return rad * c

def get_user_recommendation(n, user_config, data_config):
    # load data from csv
    data = pd.read_csv(data_config["path"])

    # load user preferences
    user_pref = user_config["preferences"]
    user_loc = user_config["location"]
    # compare with property_type (house/apartment/none)
    property_type = "" if user_pref["property_type"] is None else user_pref["property_type"]
    prop_cond = True if property_type == "" else data.property_type == property_type
    # compare with size (actual num/none)
    size = 0 if user_pref["size"] is None else user_pref["size"]
    size_cond = data['size'] >= size
    # compare with capacity (actual num/none)
    capacity = 0 if user_pref["capacity"] is None else user_pref["capacity"]
    capacity_cond = data['capacity'] >= capacity
    # compare with is_furnished (Full Furnished/Unfurnished/none)
    is_furnished = "" if user_pref["is_furnished"] is None else user_pref["is_furnished"]
    furnish_cond = True if is_furnished == "" else data['is_furnished'] == is_furnished
    # compare with yearly_price (actual num/none)
    yearly_price = 0 if user_pref["yearly_price"] is None else user_pref["yearly_price"]
    yearly_cond = data['yearly_price'] <= yearly_price
    lat = user_loc["latitude"]
    long = user_loc["longitude"]

    # get data by user preferences
    data_by_preferences = data[prop_cond & size_cond & capacity_cond & furnish_cond & yearly_cond].copy()
    data_by_preferences['distance_km'] = data_by_preferences.apply(
        lambda row: haversine(lat, long, row['latitude'], row['longitude']),
        axis=1
    )
    data_by_preferences.sort_values(by='distance_km', inplace=True)
    
    # remove distance_km column
    data_by_preferences.drop(columns='distance_km', inplace=True)
    
    return data_by_preferences.head(n)

`````
