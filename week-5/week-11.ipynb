# Study Case 2: Akses Data Penumpang Kapal (2)

## Case 2.1

```python
cond = titanic_data["Sex"] == "female"

titanic_data.loc[cond]
```

##  Case 2.2

```python
# penumpang yang memiliki usia di atas 40 tahun

age_condition = titanic_data["Age"] > 40

titanic_data[age_condition][["Name", "Age", "Pclass", "Fare"]]
```

## Case 2.3

```python
# tampilkan data penumpang yang harga tiketnya di atas $50 dan berangkat dari port Southampton

fare_cond = titanic_data["Fare"] > 50
port_cond = titanic_data["Embarked"] == "S"

titanic_data[fare_cond & port_cond]
```

## Case 2.4

```python
# Tampilkan informasi Name, Embarked, SibSp dan Parch dari  penumpang yang SibSp-nya lebih dari 6 atau Parch lebih dari 2

sibsp_cond = titanic_data["SibSp"] > 6
parch_cond = titanic_data["Parch"] > 2

titanic_data[sibsp_cond | parch_cond][["Name", "Embarked", "SibSp", "Parch"]]
```

## Exercise

1. Tampilkan nama dan gender penumpang yang survived dan memiliki  usia dibawah 25 tahun.
2. Tampilkan data penumpang yang survived dan berangkat dari port Queenstown atau Southampton
3. Tampilkan data penumpang yang jumlah anggota keluarganya lebih dari 7
4. Tampilkan data penumpang yang tidak survived, memiliki anggota keluarga lebih dari 5 dan tidak berangkat dari port Queenstown

```python
# Tampilkan nama dan gender penumpang yang survived dan memiliki  usia dibawah 25 tahun.

surv_cond = titanic_data["Survived"] == 1
age_cond = titanic_data["Age"] < 25

titanic_data[surv_cond & age_cond][["Name", "Sex"]]
```

```python
# Tampilkan data penumpang yang survived dan berangkat dari port Queenstown atau Southampton

surv_cond = titanic_data["Survived"] == 1
port_queenstown_cond = titanic_data["Embarked"] == "Q"
port_southampton_cond = titanic_data["Embarked"] == "S"

titanic_data[surv_cond & (port_queenstown_cond | port_southampton_cond)]
```

```python
# Tampilkan data penumpang yang jumlah anggota keluarganya lebih dari 7

titanic_data["FamSize"] = titanic_data["SibSp"] + titanic_data["Parch"]
total_fam_cond = titanic_data["FamSize"] > 7

titanic_data[total_fam_cond]
```

```python
# Tampilkan data penumpang yang tidak survived, memiliki anggota keluarga lebih dari 5 dan tidak berangkat dari port Queenstown

surv_cond = titanic_data["Survived"] == 0
total_fam_cond = titanic_data["FamSize"] > 5
port_cond = titanic_data["Embarked"] != "Q"

titanic_data[surv_cond & total_fam_cond & port_cond]
```

# Study Case 3: Analisis Karakter di Film

```python
lotr = pd.read_csv("WordsByCharacter.csv")

lotr.head()
```

## Case 3.1

```python
# Siapa saja karakter yang berbicara pada chapter pertama "The Fellowship of the Ring?"

talk_cond = lotr["Words"] > 0
movie_cond = lotr["Film"].str.contains("Fellowship")
chapter_cond = lotr["Chapter"] == "01: Prologue"

lotr[movie_cond & chapter_cond]
```

```python
multi = lotr.set_index(["Film", "Chapter", "Race", "Character"])

multi
```

```python
multi.loc[("The Fellowship Of The Ring", "01: Prologue"), :]
```

```python
# Siapa tiga elf pertama yang berbicara pada film "The Fellowship of the Ring"?

elf_cond = lotr["Race"] == "Elf"
film_cond = lotr["Film"] == "The Fellowship Of The Ring"

lotr[elf_cond & film_cond].set_index(["Film", "Chapter", "Race", "Character"]).head(3)
```

## Exercise

1. Siapakah karakter yang berbicara pada chapter 03: The Shire pada "The Fellowship of the Ring"
2. Berapa kali Gollum berbicara untuk tiap chapter pada "The Return Of The King".

```python
# Siapakah karakter yang berbicara pada chapter 03: The Shire pada "The Fellowship of the Ring"

chapter_cond = lotr["Chapter"] == "03: The Shire"
film_cond = lotr["Film"] == "The Fellowship Of The Ring"

lotr[chapter_cond & film_cond]
```

```python
# Berapa kali Gollum berbicara untuk tiap chapter pada "The Return Of The King".

char_cond = lotr["Character"] == "Gollum"
film_cond = lotr["Film"] == "The Return Of The King"

lotr[char_cond & film_cond]
```
