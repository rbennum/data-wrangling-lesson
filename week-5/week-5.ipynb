## Exercise

1. Ubah nilai Afendy dan Sugeng masing-masing menjadi 85.5 dan 77.77

2. Ubah data nilai tersebut menjadi rentang 1-10

3. Hapus nilai Matheos dan Fandy

4. Export data menjadi data excel


```python
# Ubah nilai Afendy dan Sugeng
data[["Afendy", "Sugeng"]] = [85.5, 77.77]
data
```

```python
# Ubah data nilai jadi rentang 1-10
def normalize(value):
    return ((value - 0) / (100 - 0)) * (10 - 1) + 1


def isnormalized(series):
    return series.between(1, 10).all()


if isnormalized(data):
    pass
else:
    data = data.map(normalize)

data
```

```python
# Hapus nilai Matheos and Fandy
if data.index.isin(["Matheos", "Fandy"]).any():
    data = data.drop(["Matheos", "Fandy"])
data
```

```python
# export data menjadi data excel
import os.path

if os.path.isfile("mahasiswa.xlsx"):
    pass
else:
    with pd.ExcelWriter("mahasiswa.xlsx") as writer:
        data.to_excel(writer, sheet_name="nilai_mahasiswa_2025")
```

## Exercise

1. Ubah masing-masing header sesuai data berikut:

2. Tambah data berikut

3. Ubah umur data index 1200 menjadi 31 dan update data index 1002 menjadi

4. Export data menjadi data excel


```python
# Ubah masing-masing header sesuai data berikut
data = data.rename(
    columns={
        "Genre": "JenisKelamin",
        "Age": "Umur",
        "Annual_Income_(k$)": "Pendapatan",
        "Spending_Score": "Skor",
    }
)
data
```

```python
row1 = pd.DataFrame(
    {"JenisKelamin": "Male", "Umur": 23, "Pendapatan": 100.5, "Skor": 50.4},
    index=[1202],
)
row2 = pd.DataFrame(
    {"JenisKelamin": "Female", "Umur": 45, "Pendapatan": 70.5, "Skor": 35.6},
    index=[1205],
)
row3 = pd.DataFrame(
    {"JenisKelamin": "Male", "Umur": 35, "Pendapatan": 88.3, "Skor": 54.8}, index=[1207]
)
data = pd.concat([data, row1, row2, row3])
data
```

```python
data.at[1200, "Umur"] = 31

data.loc[[1200], :]
```

```python
data.loc[[1002], :] = pd.DataFrame(
    {"JenisKelamin": "Male", "Umur": 22, "Pendapatan": 20, "Skor": 83}, index=[1002]
)
data.loc[[1002], :]
```

```python
with pd.ExcelWriter("mall_customer.xlsx") as writer:
    data.to_excel(writer, sheet_name="mall_customers_2025")
```
